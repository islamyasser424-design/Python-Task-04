```python
# =====================================
# Python Practice - Q2
# =====================================

# Write a function called calculate_average(numbers).
#
# The function should:
# 1. Receive a list of numbers.
# 2. Calculate the average using a loop.
# 3. Return the average.
# 4. Return 0 if the list is empty.
#
# Do not use sum().
#
# Example:
# calculate_average([10, 20, 30, 40])
# Expected output: 25.0
```

```python
def calculate_average(numbers):
    if len(numbers) == 0:
        return 0
    total = 0
    for number in numbers:
        total += number
    return total / len(numbers)

print(calculate_average([10, 20, 30, 40]))
```

```python
# =====================================
# Python Practice - Q4
# =====================================

# Write a function called find_largest(numbers).
#
# The function should return the largest number
# in a list without using max() or sort().
#
# If the list is empty, return None.
#
# Example:
# find_largest([12, 45, 7, 89, 23])
# Expected output: 89
```

```python
def find_largest(numbers):
    if len(numbers) == 0:
        return None
    largest = numbers[0]
    for number in numbers:
        if number > largest:
            largest = number
    return largest

print(find_largest([12, 45, 7, 89, 23]))
```

```python
# =====================================
# Python Practice - Q5
# =====================================

# Write a function called remove_duplicates(numbers).
#
# The function should return a new list containing
# unique values while preserving the original order.
#
# Do not convert the list into a set.
#
# Example:
# remove_duplicates([10, 20, 10, 30, 20, 40])
# Expected output: [10, 20, 30, 40]
```

```python
def remove_duplicates(numbers):
    new_list = []
    for number in numbers:
        if number not in new_list:
            new_list.append(number)
    return new_list

print(remove_duplicates([10, 20, 10, 30, 20, 40]))
```

```python
# =====================================
# Python Practice - Q6
# =====================================

students = {
    "Ahmed": 85,
    "Sara": 92,
    "Ali": 67,
    "Mona": 78,
    "Omar": 95
}

# Write a function called analyze_grades(students).
#
# The function should return a dictionary containing:
# 1. The average grade.
# 2. The highest grade.
# 3. The name of the highest-scoring student.
# 4. The number of students who scored 80 or higher.
#
# Do not use max() or sum().
```

```python
def analyze_grades(students):
    if len(students) == 0:
        return {"average": 0, "highest_grade": None, "top_student": None, "above_80": 0}
    total = 0
    highest = -1
    top_student = ""
    above_80 = 0
    for name in students:
        grade = students[name]
        total += grade
        if grade > highest:
            highest = grade
            top_student = name
        if grade >= 80:
            above_80 += 1
    return {"average": total / len(students), "highest_grade": highest, "top_student": top_student, "above_80": above_80}

print(analyze_grades(students))
```

```python
# =====================================
# Python Practice - Q7
# =====================================

# Write a function called validate_password(password).
#
# A valid password must:
# 1. Contain at least 8 characters.
# 2. Contain at least one uppercase letter.
# 3. Contain at least one lowercase letter.
# 4. Contain at least one digit.
# 5. Contain at least one special character.
#
# Return True if all conditions are satisfied,
# otherwise return False.
#
# Do not use regular expressions.
```

```python
def validate_password(password):
    upper = False
    lower = False
    digit = False
    special = False
    for char in password:
        if char.isupper():
            upper = True
        elif char.islower():
            lower = True
        elif char.isdigit():
            digit = True
        else:
            special = True
    return len(password) >= 8 and upper and lower and digit and special

print(validate_password("Hello123!"))
print(validate_password("hello123"))
```

```python
# =====================================
# Python Practice - Q8
# =====================================

# Write a function called merge_sales(sales1, sales2).
#
# Both dictionaries contain product names as keys
# and sales values as values.
#
# If a product exists in both dictionaries,
# add its sales values together.
#
# If a product exists in only one dictionary,
# keep its value unchanged.
#
# Return a new dictionary without modifying
# the original dictionaries.
#
# Example:
# sales1 = {"Laptop": 1000, "Phone": 500}
# sales2 = {"Phone": 300, "Tablet": 700}
#
# Expected output:
# {"Laptop": 1000, "Phone": 800, "Tablet": 700}
```

```python
def merge_sales(sales1, sales2):
    result = sales1.copy()
    for product in sales2:
        if product in result:
            result[product] += sales2[product]
        else:
            result[product] = sales2[product]
    return result

sales1 = {"Laptop": 1000, "Phone": 500}
sales2 = {"Phone": 300, "Tablet": 700}
print(merge_sales(sales1, sales2))
```

```python
# =====================================
# Python Practice - Q10
# =====================================

orders = [
    {"product": "Laptop", "quantity": 2, "price": 25000},
    {"product": "Phone", "quantity": 3, "price": 15000},
    {"product": "Laptop", "quantity": 1, "price": 25000},
    {"product": "Tablet", "quantity": 4, "price": 10000},
    {"product": "Phone", "quantity": 2, "price": 15000}
]

# Create the following functions:
#
# 1. calculate_order_total(order)
#    Returns quantity * price for one order.
#
# 2. calculate_total_revenue(orders)
#    Returns the total revenue of all orders.
#
# 3. calculate_product_revenue(orders)
#    Returns a dictionary containing total revenue
#    for each product.
#
# 4. find_best_selling_product(orders)
#    Returns the product with the highest total revenue.
#
# Use the first function inside the other functions
# whenever appropriate.
```

```python
def calculate_order_total(order):
    return order["quantity"] * order["price"]

def calculate_total_revenue(orders):
    total = 0
    for order in orders:
        total += calculate_order_total(order)
    return total

def calculate_product_revenue(orders):
    result = {}
    for order in orders:
        product = order["product"]
        if product not in result:
            result[product] = 0
        result[product] += calculate_order_total(order)
    return result

def find_best_selling_product(orders):
    revenues = calculate_product_revenue(orders)
    best = None
    highest = -1
    for product in revenues:
        if revenues[product] > highest:
            highest = revenues[product]
            best = product
    return best

print(calculate_total_revenue(orders))
print(calculate_product_revenue(orders))
print(find_best_selling_product(orders))
```

```python
# =====================================
# Python Practice - Q11
# =====================================

text = "Python is easy and Python is powerful and easy"

# Create the following functions:
#
# 1. clean_text(text)
#    Converts the text to lowercase and removes
#    basic punctuation such as commas and periods.
#
# 2. count_words(text)
#    Returns a dictionary containing each word
#    and its frequency.
#
# 3. find_most_common_word(text)
#    Returns the word with the highest frequency.
#
# 4. get_unique_words(text)
#    Returns a list of unique words.
#
# Reuse the functions wherever appropriate.
```

```python
def clean_text(text):
    text = text.lower()
    for mark in ",.!?;:":
        text = text.replace(mark, "")
    return text

def count_words(text):
    words = clean_text(text).split()
    counts = {}
    for word in words:
        if word in counts:
            counts[word] += 1
        else:
            counts[word] = 1
    return counts

def find_most_common_word(text):
    counts = count_words(text)
    common = None
    biggest = 0
    for word in counts:
        if counts[word] > biggest:
            biggest = counts[word]
            common = word
    return common

def get_unique_words(text):
    words = clean_text(text).split()
    unique = []
    for word in words:
        if word not in unique:
            unique.append(word)
    return unique

print(count_words(text))
print(find_most_common_word(text))
print(get_unique_words(text))
```

```python
# =====================================
# Python Practice - Q12
# =====================================

employees = {
    101: {"name": "Ahmed", "department": "IT", "salary": 15000},
    102: {"name": "Sara", "department": "HR", "salary": 18000},
    103: {"name": "Ali", "department": "IT", "salary": 22000}
}

# Create the following functions:
#
# 1. add_employee(employees, emp_id, name, department, salary)
#    Adds a new employee if the ID does not already exist.
#
# 2. update_salary(employees, emp_id, new_salary)
#    Updates the salary if the employee exists.
#
# 3. get_department_employees(employees, department)
#    Returns the names of employees in the given department.
#
# 4. calculate_average_salary(employees)
#    Returns the average salary of all employees.
#
# 5. find_highest_paid_employee(employees)
#    Returns the name and salary of the highest-paid employee.
#
# Handle invalid employee IDs and empty dictionaries.
```

```python
def add_employee(employees, emp_id, name, department, salary):
    if emp_id in employees or salary < 0:
        return False
    employees[emp_id] = {"name": name, "department": department, "salary": salary}
    return True

def update_salary(employees, emp_id, new_salary):
    if emp_id not in employees or new_salary < 0:
        return False
    employees[emp_id]["salary"] = new_salary
    return True

def get_department_employees(employees, department):
    names = []
    for emp_id in employees:
        if employees[emp_id]["department"] == department:
            names.append(employees[emp_id]["name"])
    return names

def calculate_average_salary(employees):
    if len(employees) == 0:
        return 0
    total = 0
    for emp_id in employees:
        total += employees[emp_id]["salary"]
    return total / len(employees)

def find_highest_paid_employee(employees):
    if len(employees) == 0:
        return None
    name = None
    highest = -1
    for emp_id in employees:
        employee = employees[emp_id]
        if employee["salary"] > highest:
            highest = employee["salary"]
            name = employee["name"]
    return (name, highest)

print(get_department_employees(employees, "IT"))
print(calculate_average_salary(employees))
print(find_highest_paid_employee(employees))
```

```python
# =====================================
# Python Practice - Q13
# =====================================

products = {
    "Laptop": 25000,
    "Phone": 15000,
    "Tablet": 10000,
    "Headphones": 2000
}

cart = {
    "Laptop": 1,
    "Phone": 2,
    "Headphones": 3
}

# Create the following functions:
#
# 1. calculate_cart_total(cart, products)
#    Calculates the total price of all items.
#
# 2. apply_discount(total, discount_percentage)
#    Returns the total after applying a discount.
#
# 3. count_cart_items(cart)
#    Returns the total number of individual items.
#
# 4. get_expensive_products(cart, products, threshold)
#    Returns the products whose unit price is greater
#    than the given threshold.
#
# 5. checkout(cart, products, discount_percentage)
#    Returns a dictionary containing the subtotal,
#    discount amount, and final total.
#
# Validate that quantities are positive and
# discount percentages are between 0 and 100.
```

```python
def calculate_cart_total(cart, products):
    total = 0
    for product in cart:
        if product not in products or cart[product] <= 0:
            return None
        total += products[product] * cart[product]
    return total

def apply_discount(total, discount_percentage):
    if total is None or discount_percentage < 0 or discount_percentage > 100:
        return None
    return total - total * discount_percentage / 100

def count_cart_items(cart):
    count = 0
    for product in cart:
        if cart[product] <= 0:
            return None
        count += cart[product]
    return count

def get_expensive_products(cart, products, threshold):
    result = []
    for product in cart:
        if product in products and cart[product] > 0 and products[product] > threshold:
            result.append(product)
    return result

def checkout(cart, products, discount_percentage):
    subtotal = calculate_cart_total(cart, products)
    final = apply_discount(subtotal, discount_percentage)
    if final is None:
        return None
    return {"subtotal": subtotal, "discount": subtotal - final, "final_total": final}

print(count_cart_items(cart))
print(get_expensive_products(cart, products, 10000))
print(checkout(cart, products, 10))
```

```python
# =====================================
# Python Practice - Q14
# =====================================

students = [
    {"name": "Ahmed", "grades": [85, 90, 78]},
    {"name": "Sara", "grades": [95, 92, 98]},
    {"name": "Ali", "grades": [65, 70, 72]},
    {"name": "Mona", "grades": [88, 85, 91]},
    {"name": "Omar", "grades": [75, 80, 77]}
]

# Create the following functions:
#
# 1. calculate_student_average(student)
#    Returns the average grade of one student.
#
# 2. assign_grade(average)
#    Returns:
#    "A" for average >= 90
#    "B" for average >= 80
#    "C" for average >= 70
#    "D" for average >= 60
#    "F" otherwise.
#
# 3. generate_report(students)
#    Returns a new list of dictionaries containing:
#    name, average, and letter grade.
#
# 4. get_top_students(report, threshold)
#    Returns the names of students whose averages
#    are greater than or equal to the threshold.
#
# 5. rank_students(report)
#    Returns students sorted by average, descending.
#
# 6. calculate_class_statistics(report)
#    Returns the class average and highest average.
#
# Reuse your functions instead of repeating code.
# Handle students whose grades list is empty.
```

```python
def calculate_student_average(student):
    grades = student["grades"]
    if len(grades) == 0:
        return 0
    total = 0
    for grade in grades:
        total += grade
    return total / len(grades)

def assign_grade(average):
    if average >= 90:
        return "A"
    elif average >= 80:
        return "B"
    elif average >= 70:
        return "C"
    elif average >= 60:
        return "D"
    else:
        return "F"

def generate_report(students):
    report = []
    for student in students:
        average = calculate_student_average(student)
        report.append({"name": student["name"], "average": average, "grade": assign_grade(average)})
    return report

def get_top_students(report, threshold):
    names = []
    for student in report:
        if student["average"] >= threshold:
            names.append(student["name"])
    return names

def rank_students(report):
    ranked = report.copy()
    for i in range(len(ranked)):
        for j in range(i + 1, len(ranked)):
            if ranked[j]["average"] > ranked[i]["average"]:
                ranked[i], ranked[j] = ranked[j], ranked[i]
    return ranked

def calculate_class_statistics(report):
    if len(report) == 0:
        return {"class_average": 0, "highest_average": 0}
    total = 0
    highest = 0
    for student in report:
        average = student["average"]
        total += average
        if average > highest:
            highest = average
    return {"class_average": total / len(report), "highest_average": highest}

report = generate_report(students)
print(report)
print(get_top_students(report, 85))
print(rank_students(report))
print(calculate_class_statistics(report))
```

```python
# =====================================
# Python Practice - Q15
# =====================================

accounts = {
    1001: {"name": "Ahmed", "balance": 5000},
    1002: {"name": "Sara", "balance": 8000},
    1003: {"name": "Ali", "balance": 3000}
}

# Create the following functions:
#
# 1. check_balance(accounts, account_id)
#    Returns the current balance.
#
# 2. deposit(accounts, account_id, amount)
#    Adds money if the account exists and amount > 0.
#
# 3. withdraw(accounts, account_id, amount)
#    Withdraws money only if the amount is positive
#    and sufficient funds are available.
#
# 4. transfer(accounts, sender_id, receiver_id, amount)
#    Transfers money between two valid accounts.
#    Ensure the transfer is completed only if all
#    validations pass.
#
# 5. get_richest_customer(accounts)
#    Returns the name of the customer with the
#    highest balance.
#
# 6. bank_report(accounts)
#    Returns the total money held by the bank,
#    the average account balance, and the number
#    of accounts.
#
# 7. main()
#    Displays a menu that lets the user:
#    - Check balance
#    - Deposit money
#    - Withdraw money
#    - Transfer money
#    - Display bank report
#    - Exit
#
# Use a loop to keep the program running until
# the user selects Exit.
#
# Handle invalid account IDs, invalid amounts,
# and insufficient balances.
```

```python
def check_balance(accounts, account_id):
    if account_id not in accounts:
        return None
    return accounts[account_id]["balance"]

def deposit(accounts, account_id, amount):
    if account_id not in accounts or amount <= 0:
        return False
    accounts[account_id]["balance"] += amount
    return True

def withdraw(accounts, account_id, amount):
    if account_id not in accounts or amount <= 0:
        return False
    if accounts[account_id]["balance"] < amount:
        return False
    accounts[account_id]["balance"] -= amount
    return True

def transfer(accounts, sender_id, receiver_id, amount):
    if sender_id not in accounts or receiver_id not in accounts:
        return False
    if sender_id == receiver_id or amount <= 0:
        return False
    if accounts[sender_id]["balance"] < amount:
        return False
    withdraw(accounts, sender_id, amount)
    deposit(accounts, receiver_id, amount)
    return True

def get_richest_customer(accounts):
    richest = None
    highest = -1
    for account_id in accounts:
        if accounts[account_id]["balance"] > highest:
            highest = accounts[account_id]["balance"]
            richest = accounts[account_id]["name"]
    return richest

def bank_report(accounts):
    total = 0
    for account_id in accounts:
        total += accounts[account_id]["balance"]
    count = len(accounts)
    average = 0
    if count > 0:
        average = total / count
    return {"total_money": total, "average_balance": average, "accounts_count": count}

def main():
    while True:
        print("1. Check balance")
        print("2. Deposit")
        print("3. Withdraw")
        print("4. Transfer")
        print("5. Bank report")
        print("6. Exit")
        choice = input("Choose: ")
        if choice == "6":
            break
        elif choice == "5":
            print(bank_report(accounts))
        elif choice in ["1", "2", "3", "4"]:
            try:
                account_id = int(input("Account ID: "))
                if choice == "1":
                    print(check_balance(accounts, account_id))
                elif choice == "2":
                    amount = float(input("Amount: "))
                    print(deposit(accounts, account_id, amount))
                elif choice == "3":
                    amount = float(input("Amount: "))
                    print(withdraw(accounts, account_id, amount))
                elif choice == "4":
                    receiver = int(input("Receiver ID: "))
                    amount = float(input("Amount: "))
                    print(transfer(accounts, account_id, receiver, amount))
            except ValueError:
                print("Invalid number")
        else:
            print("Invalid choice")
```
